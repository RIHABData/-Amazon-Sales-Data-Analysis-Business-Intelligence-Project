# Import Librairies

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 🎧 STEP 1: Load Data

```python
Sales = pd.read_csv('amazon.csv')
Sales
```

```python
# Display the first few rows of the dataset
Sales.head()
```

## 🧹 STEP 2: Data Cleaning

```python
# Information of the dataset
Sales.info()
```

```python
# Summary statistics using .describe()
Sales.describe(include='all')
```

```python
# Checking if missing data or null values are present in the dataset
Sales.isnull().sum()
```

```python
# Remove duplicates
Sales.drop_duplicates(inplace=True)
```

```python
# Handle missing values
Sales.fillna(0, inplace=True)
```

## 🧠 STEP 3: Feature Engineering

```python
Sales.columns
```

```python
Sales.dtypes
```

```python
Sales['actual_price'] = Sales['actual_price'].replace('[^0-9.]', '', regex=True)
Sales['discounted_price'] = Sales['discounted_price'].replace('[^0-9.]', '', regex=True)
```

```python
Sales['actual_price'] = pd.to_numeric(Sales['actual_price'], errors='coerce')
Sales['discounted_price'] = pd.to_numeric(Sales['discounted_price'], errors='coerce')
```

```python
# 💰 Discount Amount
Sales['discount_amount'] = Sales['actual_price'] - Sales['discounted_price']
```

```python
# 📉 Price Ratio
Sales['price_ratio'] = Sales['discounted_price'] / Sales['actual_price']
```

```python
# 📊 Convert rating to numeric
Sales['rating'] = pd.to_numeric(Sales['rating'], errors='coerce')
```

## 📊 STEP 4: EDA 

```python
# 📊 Distribution of prices
Sales['actual_price'].hist(bins=50)
plt.title("Actual Price Distribution")
plt.show()
```

```python
# 💸 Discounted vs Actual Price
plt.scatter(Sales['actual_price'], Sales['discounted_price'])
plt.xlabel("Actual Price")
plt.ylabel("Discounted Price")
plt.title("Price Comparison")
plt.show()
```

```python
# 📊 Rating distribution
Sales['rating'].hist(bins=20)
plt.title("Rating Distribution")
plt.show()
```

```python
# 📊 Discount vs Rating
plt.scatter(Sales['discount_percentage'], Sales['rating'])
plt.xlabel("Discount %")
plt.ylabel("Rating")
plt.title("Discount vs Rating")
plt.show()
```

```python
# 🥧 Pie Chart Using category
category_counts = Sales['category'].value_counts().head(5)

plt.figure(figsize=(8,8))

plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct='%1.1f%%'
)

plt.title("Top 5 Categories")

plt.show()
```

```python
# 🍩 Donut Chart Using category

category_counts = Sales['category'].value_counts().head(5)

plt.figure(figsize=(8,8))

plt.pie(
    category_counts.values,
    labels=category_counts.index,
    autopct='%1.1f%%'
)

# donut hole
centre_circle = plt.Circle((0,0),0.70,fc='white')

fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Top Categories Donut Chart")

plt.show()
```

## WordCloud

```python

from wordcloud import WordCloud

# Remove missing values
Sales = Sales.dropna(subset=['category'])

# Count category frequency
category_freq = Sales['category'].value_counts().to_dict()

# Generate WordCloud
wordcloud = WordCloud(
    width=1200,
    height=600,
    background_color='white',
    colormap='viridis'
).generate_from_frequencies(category_freq)

# Plot
plt.figure(figsize=(14,7))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Amazon Product Categories WordCloud")
plt.show()
```
